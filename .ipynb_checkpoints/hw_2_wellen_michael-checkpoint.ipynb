```python
%%file sum_distance.py
from mrjob.job import MRJob

class CALCULATE_SUM(MRJob):

    def mapper(self, _, line):
        (location, date, type, Route_Numb, distributi, dist, z) = line.split(',')
        if (type != 'Route_Type'):
            quest = float(dist)
            yield date, quest

    def reducer(self, date, distances):
        yield date, sum(distances)


if __name__ == '__main__':
    CALCULATE_SUM.run()
```

```python
import sum_distance

mr_job = sum_distance.CALCULATE_SUM(args=['Bus_Routes.csv'])
with mr_job.make_runner() as runner:
    runner.run()
    for key, value in mr_job.parse_output(runner.cat_output()):
        print(key, value)
```
